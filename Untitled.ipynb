```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
d = 2
x_star = np.array([0, 0])  # true optimum
lam = 10          # number of offspring per iteration
sigma = 0.5       # step size
max_iter = 50     # number of iterations
m = np.array([4, 3])  # starting point m_0
mu = 3
weights = np.ones(mu) / mu

def f(x):
    """Quadratic objective: f(x) = ||x - x*||^2"""
    return np.sum((x)**2)
def f1(x):
    return x[0]**2+100*x[1]**2
```

```python
history = [m.copy()]
f_history = [f(m)]

rng = np.random.default_rng(42)  # for reproducibility

for n in range(max_iter):
    Z = rng.standard_normal((lam, d))
    X = m + sigma * Z
    
    vals = np.array([f(X[i]) for i in range(lam)])
    
    i_best = int(np.argmin(vals))
    m = X[i_best]  # m_{n+1} = best candidate
    
    history.append(m.copy())
    f_history.append(f(m))

history = np.array(history)
f_history = np.array(f_history)


fig, ax = plt.subplots(figsize=(6, 6))

# Grid for contours
lim = 5.5
x_range = np.linspace(-lim, lim, 400)
X1, X2 = np.meshgrid(x_range, x_range)
F = (X1 - x_star[0])**2 + (X2 - x_star[1])**2

# Concentric circle contour lines matching the target figure
radii = np.array([0.5, 0.9, 1.3, 1.9, 2.9, 4.3, 6.2])
levels = radii**2
ax.contour(X1, X2, F, levels=levels, cmap='viridis', alpha=0.6, linewidths=1.2)

# ES Trajectory
ax.plot(history[:, 0], history[:, 1], '.-', color='#1f77b4', 
        label='Isotropic ES', markersize=6, linewidth=1.5)

# Start and Optimum markers
ax.plot(x_star[0], x_star[1], 'r*', markersize=16, label=r'Optimum $(0, 0)$', zorder=5)
ax.plot(history[0, 0], history[0, 1], '^', color='black', markersize=10, label=r'Start $x_0$', zorder=5)

# Formatting
ax.set_xlim(-lim, lim)
ax.set_ylim(-lim, lim)
ax.set_aspect('equal')
ax.set_xlabel(r'$x_1$', fontsize=12)
ax.set_ylabel(r'$x_2$', fontsize=12)
ax.set_title(r'2D Search Trajectory on Sphere $f(x) = \|x\|^2$', fontsize=12)
ax.grid(True, linestyle='--', alpha=0.35)
ax.legend(loc='lower right', frameon=True, fontsize=10)

plt.tight_layout()
plt.savefig('es_matched_trajectory.png', dpi=200)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Define the anisotropic quadratic problem
# f(x) = x_1^2 + 100 * x_2^2  =>  A = [[2, 0], [0, 200]]  since f(x) = 0.5 * x^T A x
A = np.diag([2.0, 200.0])

def objective(x):
    # x shape: (N, 2) or (2,)
    return 0.5 * np.sum(x @ A * x, axis=-1)

def run_es(C, init_mean, sigma=0.5, lam=30, mu=10, max_iter=60):
    """
    (mu/mu_w, lambda)-Evolution Strategy with fixed covariance matrix C.
    """
    dim = 2
    mean = np.array(init_mean, dtype=float)
    
    # Weights for top mu candidates (logarithmic/rank weights, normalized)
    weights = np.log(mu + 0.5) - np.log(np.arange(1, mu + 1))
    weights /= np.sum(weights)
    
    # Cholesky factor for sampling: X = m + sigma * L @ Z
    L = np.linalg.cholesky(C)
    
    trajectory = [mean.copy()]
    history_f = [objective(mean)]
    
    for _ in range(max_iter):
        # Sample lambda candidates: Z ~ N(0, I)
        z = np.random.randn(lam, dim)
        candidates = mean + sigma * (z @ L.T)
        
        # Evaluate and sort by fitness
        fitness = objective(candidates)
        idx_sorted = np.argsort(fitness)
        
        # Update mean using top mu candidates
        best_candidates = candidates[idx_sorted[:mu]]
        mean = np.sum(weights[:, None] * best_candidates, axis=0)
        
        trajectory.append(mean.copy())
        history_f.append(objective(mean))
        
    return np.array(trajectory), np.array(history_f)

# Run simulations
np.random.seed(42)
x0 = [4.0, 1.5]

# Method 1: Isotropic Search (C = I)
traj_iso, loss_iso = run_es(C=np.eye(2), init_mean=x0, sigma=0.2)

# Method 2: Anisotropic Search (C aligned with curvature: C = A^-1)
C_aniso = np.linalg.inv(A)
# Scale C so its trace matches the dimension
C_aniso = C_aniso / np.trace(C_aniso) * 2.0
traj_aniso, loss_aniso = run_es(C=C_aniso, init_mean=x0, sigma=0.2)

# Visualization
plt.figure(figsize=(8,5))
plt.contour(X, Y, Z, levels=np.logspace(-1, 3, 20), cmap='viridis', alpha=0.6)
plt.plot(traj_iso[:, 0], traj_iso[:, 1], 'ro-', label='Isotropic $N(m, \sigma^2 I)$', markersize=3)
plt.plot(traj_aniso[:, 0], traj_aniso[:, 1], 'b*-', label='Anisotropic $N(m, \sigma^2 C)$', markersize=4)
plt.title("Search Trajectories on $f(x) = x_1^2 + 100x_2^2$")
plt.xlabel("$x_1$")
plt.ylabel("$x_2$")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

# plt.savefig('/home/seanglengkhe/seangleng/git-local/seams_school_2026/images/iso_vs_aniso')
plt.show()


fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

# Plot 1: Contour & Trajectories
x_grid = np.linspace(-5, 5, 400)
y_grid = np.linspace(-2, 2, 400)
X, Y = np.meshgrid(x_grid, y_grid)
Z = X**2 + 100 * Y**2

ax1.contour(X, Y, Z, levels=np.logspace(-1, 3, 20), cmap='viridis', alpha=0.6)
ax1.plot(traj_iso[:, 0], traj_iso[:, 1], 'ro-', label='Isotropic $N(m, \sigma^2 I)$', markersize=3)
ax1.plot(traj_aniso[:, 0], traj_aniso[:, 1], 'b*-', label='Anisotropic $N(m, \sigma^2 C)$', markersize=4)
ax1.set_title("Search Trajectories on $f(x) = x_1^2 + 100x_2^2$")
ax1.set_xlabel("$x_1$")
ax1.set_ylabel("$x_2$")
ax1.legend()
ax1.grid(True, linestyle="--", alpha=0.5)



# Plot 2: Convergence Comparison
ax2.semilogy(loss_iso, 'r-', label='Isotropic Search')
ax2.semilogy(loss_aniso, 'b-', label='Anisotropic Search')
ax2.set_title("Objective Value vs. Iterations")
ax2.set_xlabel("Iteration")
ax2.set_ylabel("Fitness (log scale)")
ax2.legend()
ax2.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
