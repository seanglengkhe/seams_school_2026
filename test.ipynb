```python
import numpy as np
import matplotlib.pyplot as plt

# 1. High-resolution 1D axis slice
x = np.linspace(-4, 4, 4000)

# 2. Underlying convex/parabolic bowl
f_base = 5.6 * x**2

# 3. Multi-frequency oscillations creating ruggedness & local minima
oscillations = (
    12 * np.sin(16 * x)
    + 8 * np.sin(35 * x + np.cos(5 * x))
    + 4 * np.sin(80 * x)
)

# 4. Sharp drops and spikes characteristic of benchmark landscapes
sharp_features = 6 * np.abs(np.sin(20 * x)) * np.sign(np.cos(12 * x))
spikes = np.zeros_like(x)
for loc in [-2.9, -1.5, -1.3, 0.6, 2.1, 2.7, 3.6]:
    spikes += 30 / (1 + 2500 * (x - loc)**2)

# Combine terms and bound to [0, 100]
y = np.clip(f_base + oscillations + sharp_features - spikes, 0, 100)

# 5. Plot formatting matching the reference image
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, color='blue', linewidth=1.5)

# Axis limits and labels
ax.set_xlim(-4, 4)
ax.set_ylim(0, 100)
ax.set_xlabel(r'$x_i$ (component of $x$)', fontsize=15)
ax.set_ylabel(r'function value $f(x)$', fontsize=15)

# Dotted grid and inward ticks
ax.grid(True, linestyle=':', color='gray', alpha=0.7)
ax.tick_params(direction='in', top=True, right=True)

plt.tight_layout()
plt.show()
```
